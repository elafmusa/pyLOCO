```python
# compare strengths before and after realignment

import numpy as np
import matplotlib
matplotlib.use('Tkagg')
import matplotlib.pyplot as plt

# load strengths file before realignment
def read_str_file(file):
    str_dict = {}
    quad= []
    sext= []
    oct = []
    hst = []
    vst = []
    sqp = []

    with open(file, 'r') as f:
        ls = f.readlines()
        for l in ls:
            if l[0] != '#':
                ll = l.split(':')
                name = ll[0]
                str = float(ll[1][0:-1])
    #             print(f'{name} has str {str}')
                str_dict[name] = str

    for k, v in str_dict.items():
        if 'm-q' in k:
            quad.append(v)
        # if 'm-dq' in k:
        #     quad.append(v)
        if 'm-s' in k:
            sext.append(v)
        if 'm-o' in k:
            oct.append(v)
        if 'hst' in k:
            hst.append(v)
        if 'vst' in k:
            vst.append(v)
        if 'sqp' in k:
            sqp.append(v)

    str_dict['quad'] = np.array(quad)
    str_dict['sext'] = np.array(sext)
    str_dict['oct'] = np.array(oct)
    str_dict['hst'] = np.array(hst)
    str_dict['vst'] = np.array(vst)
    str_dict['sqp'] = np.array(sqp)

    return str_dict

def read_des_str_file(file):
    str_dict = {}
    quad= []
    sext= []
    oct = []
    hst = []
    vst = []
    sqp = []

    with open(file, 'r') as f:
        ls = f.readlines()
        for l in ls:
            ll = l.split(',')
            name = ll[0]
            str = float(ll[1][0:-1])
#             print(f'{name} has str {str}')
            str_dict[name] = str

    for k, v in str_dict.items():
        if 'm-q' in k:
            quad.append(v)
        #if 'm-dq' in k:
        #    quad.append(v)
        if 'm-s' in k:
            sext.append(v)
        if 'm-o' in k:
            oct.append(v)
        if 'hst' in k:
            hst.append(v)
        if 'vst' in k:
            vst.append(v)
        if 'sqp' in k:
            sqp.append(v)

    str_dict['quad'] = np.array(quad)
    str_dict['sext'] = np.array(sext)
    str_dict['oct'] = np.array(oct)
    str_dict['hst'] = np.zeros(384)
    str_dict['vst'] = np.zeros(288)
    str_dict['sqp'] = np.zeros(288)

    return str_dict

des = read_des_str_file('DesignStrengths.csv')

# exit()

# bef = read_str_file('/machfs/MDT/2026/2026_01_18/strengthscomparison/pyLOCO_correction.ts')
bef = read_str_file('DESY_MDT_pyLOCO_C2more.ts')
# aft = read_str_file('/machfs/MDT/2026/2026_01_18/strengthscomparison/restart_after_bba_optics3_LT5DA1_cycled_70perc.ts')
aft = read_str_file('operation_after_restart.ts')
```

```python
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    nrows=2,
    figsize=(20, 5.5),
    dpi=100,
    sharex=False
)
plt.subplots_adjust(hspace=0.45)

elements = [
    ('quad', 'Normal Quadrupoles'),
    ('sqp',  'Skew Quadrupoles')
]

color_pyloco = 'tab:orange'
color_filo   = 'tab:blue'

for c, (k, title) in enumerate(elements):
    b = bef[k] - des[k]   # pyLOCO
    a = aft[k] - des[k]   # FILO

    x_b = np.arange(len(b))
    x_a = np.arange(len(a))

    # FILO (behind)
    ax[c].plot(
        x_a, a,
        color=color_filo,
        linestyle='--',
        linewidth=1.1,
        alpha=0.8,
        zorder=1,
        label=f'FILO (std = {np.std(a):.2e})'
    )

    # pyLOCO (front)
    ax[c].plot(
        x_b, b,
        color=color_pyloco,
        linestyle='-',
        linewidth=1.5,
        alpha=0.9,
        zorder=2,
        label=f'pyLOCO (std = {np.std(b):.2e})'
    )

    # Panel title
    ax[c].set_title(title, fontsize=13, pad=8)

    # Physical y-axis label
    ax[c].set_ylabel(r'$\Delta$ Strength [1/m]', fontsize=12, labelpad=16)

    ax[c].set_xlabel('Index', fontsize=11)
    ax[c].legend(loc='lower right')
    ax[c].margins(x=0)

plt.show()
```
